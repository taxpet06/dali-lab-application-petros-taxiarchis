# Choice of project

My idea for a system that can help the scientists get their work done faster was to develop an app that will act as a platform that allows scientists to train and fine tune an AI segmentation model based on their already annotated images. My idea was that I would make a simple to use and straightforward application that requires minimal computer science knowledge to use effectively and yield results. By training an AI model, that will essentially automate the manual annotation process that the scientists go through and further analysis can be done using the model's predicted masks. 

I identified a couple of tasks that I would need to tackle to achieve my goal. One task was to build a modern-looking, understandable and easy to use user interface to make it as simple for the hypothetical barnacle scientists to use my application. The significantly harder task to tackle was to build a framework to allow for model training and fine tuning. The user interface would be responsible for collecting and organizing all the different types of data (in this case, images and masks) that the scientist input to it. The user interface is also responsible for abstracting the back end model fine tuning and displaying the model's training progress as well as the results that it outputs from the segmentation. The model's framework will take in images that the scientists input into the platform and their corresponding masks that are in the form of outlines. The model's expectated output after the evaluation of an image will be similar outlines that indicate the position of barnacles in the image.

It is difficult to properly evaluate the performance of the model considering the very limited data provided. Based on my experience, the best way to evaluate the accuracy of the fine-tuned model's predictions was to have both a training and a testing dataset where the testing dataset also includes human-annotated masks that can be used as a ground truth. Having said image-mask pairs, we can test the model's output against those human-annotated masks for the testing dataset and perform intersection over union calculations to assess how close the model's predictions are to the ground truth. However, considering that our dataset only has two masks, it is best that all annotated data is used for training and an evaluation of the model is condacted in a qualitative manner since using one of our images as a testing image would deplete half our trainig data set which would probably result in worse performance. It is not practical to compute IoU metrics considering the data that was provided to me, but a testing dataset would go a long way in evaluating the model's accuracy.

# Choice of framework

For my choice of user interface I went with streamlit, mostly because I found that it was easy to use, had a modern look, it was easy to deploy the app and also because it was recommended as a useful tool in the challenge. Moreover, since I used streamlit and, upon looking it up, all computation runs on the streamlit servers I decided to use a segment anything (SAM) approach that would be fairly light and could run easily without taking up too much memory or storage. Having these considerations in mind I went with Mobile SAM which is meant to be used for lightweight applications and seemed fairly straightforward to integrate with streamlit.

# Results

When trained using 200 epochs and the defualt learning rate, the model (which as mentioned in the README file can be found under the Code subfolder) seemingly performed well with unseen_img1 but did not produce any meaningful segmentation for unseen_img2. See the segmented images below.

![Unseen 1](./ResultImages/200_unseen1.jpg "Unseen 1")
![Unseen 2](./ResultImages/200_unseen2.jpg "Unseen 2")

In my opinion, this discrepancy in performance is probably because unseen_img2 was a very zoomed-in view of the barnacles and there was no training data on such images, so the model, expecting images that are similar to the zoomed-out view and having settled on what size on average a barnacle should be, could not detect barnacles meaningfully. This would probably be fixed if similar images to unseen_img2 existed in the training dataset. Of course, while significantly better than unseen image 2, unseen image 1's segmentation results are obviously not up to par with an accurate an reliable human-made annotation yet. However, considering the limited training data, and how scalable this solution is in a scenario with more of it, I would say that the model has performed a respectable classification. Despite that, there is obviously much room for improvement.

# Evaluation of Solution and Future of the Project

I believe that this automation protocol is worth pursuing further. This is because a successful and accurate model could completely eliminate the time-consuming, manual annotation pipeline for the scientists and automate that entire process all together, thus allowing the scientists to focus on more important tasks such as their statistical analysis and inferences. It is evident that one of the biggest limitations to the algorithm's accuracy is the very little training data. Training such a model with even a slightly bigger and more diverse dataset would yield significantly more reliable result. However, that would of course mean that more manual annotations would be needed. Furthermore, manual annotations as part of a testing dataset would be extremely useful in providing model validation and assessing its accuracy, because there will be a ground truth to compare the model's predictions with.

A future development that I would suggest for this project would be the inclusion of file uploaders in the ***Testing*** tab that allow the user to upload a testing dataset. With such data, a validation step could be included and intersection over union calculations could be conducted and displayed in the metrics tab so that the model is meaningfully evaluated when it comes to its accuracy.
